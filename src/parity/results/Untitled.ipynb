```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import re
import numpy as np
from pathlib import Path
```

```python
folder = Path(
    r"C:\Users\mirna\OneDrive\Desktop\oasis_data\data processing - newest\parity\results"
)

files = [
    "parity_Nationwide_Cohort_Diabetes_d.csv",
    "parity_Nationwide_Cohort_Heart_Failure__plus__Hypertension_hf_plus_hyp.csv",
    "parity_Nationwide_Cohort_Heart_Failure_hf.csv",
    "parity_Nationwide_Cohort_Hypertension_hyp.csv",
    "parity_Nationwide_Cohort_Overall.csv",
    "parity_Nationwide_Cohort_Rural.csv",
    "parity_Nationwide_Cohort_Urban.csv",
    "parity_Nationwide_Cohort_Diabetes__plus__Heart_Failure__plus__Hypertension_d_plus_hf_plus_hyp.csv",
    "parity_Nationwide_Cohort_Diabetes__plus__Heart_Failure_d_plus_hf.csv",
    "parity_Nationwide_Cohort_Diabetes__plus__Hypertension_d_plus_hyp.csv",
]

cohort_labels = {
    "parity_Nationwide_Cohort_Diabetes_d.csv": "Diabetes",
    "parity_Nationwide_Cohort_Heart_Failure__plus__Hypertension_hf_plus_hyp.csv": "Heart Failure + Hypertension",
    "parity_Nationwide_Cohort_Heart_Failure_hf.csv": "Heart Failure",
    "parity_Nationwide_Cohort_Hypertension_hyp.csv": "Hypertension",
    "parity_Nationwide_Cohort_Overall.csv": "Overall",
    "parity_Nationwide_Cohort_Rural.csv": "Rural",
    "parity_Nationwide_Cohort_Urban.csv": "Urban",
    "parity_Nationwide_Cohort_Diabetes__plus__Heart_Failure__plus__Hypertension_d_plus_hf_plus_hyp.csv": "Diabetes + Heart Failure + Hypertension",
    "parity_Nationwide_Cohort_Diabetes__plus__Heart_Failure_d_plus_hf.csv": "Diabetes + Heart Failure",
    "parity_Nationwide_Cohort_Diabetes__plus__Hypertension_d_plus_hyp.csv": "Diabetes + Hypertension",
}

combined = []

for filename in files:
    df = pd.read_csv(folder / filename)
    df["Cohort"] = "Nationwide"
    df["Subgroup"] = cohort_labels[filename]
    df["Source_File"] = filename
    combined.append(df)

combined_df = pd.concat(combined, ignore_index=True, sort=False)

output_file = folder / "parity_Nationwide_Cohort_ALL_COMBINED.csv"
combined_df.to_csv(output_file, index=False)

print(f"Saved: {output_file}")
print(f"Rows: {len(combined_df):,}")
print(f"Columns: {len(combined_df.columns):,}")
print(combined_df.groupby("Subgroup").size())
```

```python
files = [
    "parity_Texas_Cohort_Overall.csv",
    "parity_Texas_Cohort_Rural.csv",
    "parity_Texas_Cohort_Urban.csv",
    "parity_Texas_Cohort_Diabetes__plus__Heart_Failure__plus__Hypertension_d_plus_hf_plus_hyp.csv",
    "parity_Texas_Cohort_Diabetes__plus__Heart_Failure_d_plus_hf.csv",
    "parity_Texas_Cohort_Diabetes__plus__Hypertension_d_plus_hyp.csv",
    "parity_Texas_Cohort_Diabetes_d.csv",
    "parity_Texas_Cohort_Heart_Failure__plus__Hypertension_hf_plus_hyp.csv",
    "parity_Texas_Cohort_Heart_Failure_hf.csv",
    "parity_Texas_Cohort_Hypertension_hyp.csv",
]

subgroup_labels = {
    "parity_Texas_Cohort_Overall.csv": "Overall",
    "parity_Texas_Cohort_Rural.csv": "Rural",
    "parity_Texas_Cohort_Urban.csv": "Urban",
    "parity_Texas_Cohort_Diabetes__plus__Heart_Failure__plus__Hypertension_d_plus_hf_plus_hyp.csv":
        "Diabetes + Heart Failure + Hypertension",
    "parity_Texas_Cohort_Diabetes__plus__Heart_Failure_d_plus_hf.csv":
        "Diabetes + Heart Failure",
    "parity_Texas_Cohort_Diabetes__plus__Hypertension_d_plus_hyp.csv":
        "Diabetes + Hypertension",
    "parity_Texas_Cohort_Diabetes_d.csv": "Diabetes",
    "parity_Texas_Cohort_Heart_Failure__plus__Hypertension_hf_plus_hyp.csv":
        "Heart Failure + Hypertension",
    "parity_Texas_Cohort_Heart_Failure_hf.csv": "Heart Failure",
    "parity_Texas_Cohort_Hypertension_hyp.csv": "Hypertension",
}

dataframes = []

for filename in files:
    file_path = folder / filename

    if not file_path.exists():
        print(f"WARNING: File not found: {file_path}")
        continue

    df = pd.read_csv(file_path)

    # Preserve provenance and create a clean, consistent grouping field.
    df["Analysis_Cohort"] = "Texas"
    df["Analysis_Subgroup"] = subgroup_labels[filename]
    df["Source_File"] = filename

    dataframes.append(df)

combined_df = pd.concat(dataframes, ignore_index=True, sort=False)

# Optional: move the new identification columns to the beginning.
id_columns = ["Analysis_Cohort", "Analysis_Subgroup", "Source_File"]
other_columns = [col for col in combined_df.columns if col not in id_columns]
combined_df = combined_df[id_columns + other_columns]

output_file = folder / "parity_Texas_Cohort_ALL_COMBINED.csv"
combined_df.to_csv(output_file, index=False)

print(f"\nSaved combined file to:\n{output_file}")
print(f"\nTotal rows: {len(combined_df):,}")
print(f"Total columns: {len(combined_df.columns):,}")
print("\nRecords by subgroup:")
print(combined_df.groupby("Analysis_Subgroup").size().sort_index())
```
