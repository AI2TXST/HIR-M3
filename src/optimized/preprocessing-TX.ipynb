```python
cat_cols = ['Submitted_HIPPS_Code',
 'Facility_Internal_ID',
 'Gender',
 'ByDiscipline',
 'Agency_Medicare_Number',
 'BMI_Category',
 'Primary_Diagnosis_ICD_10_C_M_Code_Cluster',
 'Other_Diagnosis_Code_1_ICD_10_C_M_Cluster',
 'Other_Diagnosis_Code_2_ICD_10_C_M_Cluster',
 'Other_Diagnosis_Code_3_ICD_10_C_M_Cluster',
 'Other_Diagnosis_Code_4_ICD_10_C_M_Cluster',
 'Other_Diagnosis_Code_5_ICD_10_C_M_Cluster',
 'COUNTY_NAME']

cat_force_cols = [
    'Facility_Internal_ID',
    'Agency_Medicare_Number',
    'Primary_Diagnosis_ICD_10_C_M_Code_Cluster',
    'Other_Diagnosis_Code_1_ICD_10_C_M_Cluster',
    'Other_Diagnosis_Code_2_ICD_10_C_M_Cluster',
    'Other_Diagnosis_Code_3_ICD_10_C_M_Cluster',
    'Other_Diagnosis_Code_4_ICD_10_C_M_Cluster',
    'Other_Diagnosis_Code_5_ICD_10_C_M_Cluster',
]

for c in cat_force_cols:
    if c in df_model.columns:
        df_model[c] = df_model[c].astype('category')

# One‑hot encode only the categorical columns
df_cat_encoded = pd.get_dummies(
    df_model[cat_cols],
    drop_first=True, dtype=int
)

# Drop original categorical columns from df_model
df_model = df_model.drop(columns=cat_cols, errors="ignore")

# Concatenate numeric, flag, and encoded categorical columns
df_model = pd.concat([df_model, df_cat_encoded], axis=1)
df_model
```

```python
target = "ever_readmitted"

# Drop Beneficiary_ID and non-numeric; corr uses only numeric columns by default
df_corr = df_model.drop(columns=["Beneficiary_ID"], errors="ignore")

corr_with_target = df_corr.corr(numeric_only=True)[target].drop(labels=[target])
corr_with_target_sorted = corr_with_target.reindex(
    corr_with_target.abs().sort_values(ascending=False).index
)

threshold = 0.6
high_corr = corr_with_target_sorted[ corr_with_target_sorted.abs() > threshold ]
high_corr
```

```python
# Make a copy to avoid side effects
df_corr = df_model.copy()

# Drop Beneficiary_ID if present
df_corr = df_corr.drop(columns=["Beneficiary_ID"], errors="ignore")

# Ensure only numeric columns are used (corr ignores non-numeric by default)
corr_matrix = df_corr.corr(numeric_only=True)
corr_matrix
```

```python
import numpy as np

corr_abs = corr_matrix.abs()

threshold = 0.6

# Take upper triangle (excluding diagonal)
upper = corr_abs.where(np.triu(np.ones(corr_abs.shape), k=1).astype(bool))

# Columns to drop: any column that has a correlation > threshold with some earlier column
to_drop = [col for col in upper.columns if any(upper[col] > threshold)]

print("Dropping highly correlated columns:", to_drop)

# Create a reduced model dataframe
df_model_reduced = df_model.drop(columns=to_drop)

df_model_reduced
```

```python
df_model_reduced.to_csv('data/processed_final_mergedDF_TX.csv', index=False)
```
