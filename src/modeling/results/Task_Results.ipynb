### Task 1 Research Question

> How do conventional machine-learning models and transformer-based models perform for predicting the study outcome across Nationwide and Texas cohorts, Rural and Urban populations, and clinically defined condition subgroups, as evaluated by discrimination, positive-case detection, and false-positive control?

### Task 1 Objective

The objective of Task 1 was to benchmark candidate prediction models—including CatBoost, LightGBM, XGBoost, Gradient Boosting, Logistic Regression, Random Forest, Standard MLP, Standard Transformer, SAINT Transformer, and HIR-M3 Transformer—across overall, geographic, Rural–Urban, and condition-defined cohorts.

Performance was assessed using accuracy, Brier score, F1 score, PR AUC, precision, ROC AUC, recall/sensitivity, and specificity. The analysis aimed to identify:

1. The models with the strongest overall discrimination and balanced performance.
2. Whether model performance differs between Nationwide and Texas cohorts.
3. Whether performance differs between Rural and Urban populations.
4. Whether performance changes across Diabetes, Heart Failure, Hypertension, and multimorbidity cohorts.
5. Whether any model exhibits a clinically important recall-specificity trade-off.
6. The best conventional base models and complementary high-recall models to carry forward into Task 2 ensemble-ratio evaluation.

## Nationwide vs. Texas Performance

Overall, conventional machine-learning models—particularly CatBoost, LightGBM, XGBoost, and Logistic Regression—performed consistently well across both the Nationwide and Texas cohorts. ROC AUC values were broadly similar between cohorts, generally ranging from approximately 0.80 to 0.83 among the strongest models, indicating stable discrimination between positive and negative cases.

The Nationwide cohort generally showed higher accuracy and specificity. For example, CatBoost achieved an accuracy of 0.815 in the Nationwide cohort compared with 0.798 in the Texas cohort, while specificity was 0.849 versus 0.829. This pattern suggests that nationwide models were somewhat better at correctly identifying negative cases and limiting false positives.

In contrast, Texas models often showed stronger positive-class performance, including higher F1 score, PR AUC, precision, and recall. For example, the best Texas PR AUC was 0.484 for XGBoost, compared with 0.424 for the full Nationwide cohort. These results suggest that models trained or evaluated in Texas may be relatively better at identifying positive cases, while nationwide models may prioritize overall classification accuracy and specificity.

## Rural vs. Urban Performance

### Nationwide Cohort

Within the Nationwide cohort, urban performance was generally higher for accuracy, ROC AUC, and specificity among conventional models. For example, CatBoost achieved an accuracy of 0.820 in the urban subgroup compared with 0.813 in the rural subgroup. CatBoost also had slightly higher ROC AUC in urban patients (0.827) than rural patients (0.822).

However, rural Nationwide results were often stronger for metrics focused on positive-case prediction. For CatBoost, F1 score increased from 0.443 in the urban cohort to 0.467 in the rural cohort, PR AUC increased from 0.413 to 0.445, and precision increased from 0.356 to 0.391. This suggests that the rural Nationwide cohort may support better identification of positive cases, even though urban performance is somewhat stronger for overall accuracy and negative-case classification.

### Texas Cohort

The Texas rural-versus-urban comparison was less uniform. Urban Texas results had higher accuracy for several models, including CatBoost, Gradient Boosting, and XGBoost. For example, XGBoost accuracy was 0.782 in urban Texas compared with 0.703 in rural Texas.

Rural Texas frequently showed stronger recall and, for several models, stronger F1 score and PR AUC. Gradient Boosting had recall of 0.759 in rural Texas versus 0.638 in urban Texas, while its F1 score was 0.473 rural versus 0.456 urban. This indicates that rural Texas models may identify a greater share of true positive cases.

Urban Texas generally had higher specificity. For example, XGBoost specificity was 0.816 in urban Texas versus 0.690 in rural Texas, and Gradient Boosting specificity was 0.804 versus 0.746. Thus, urban Texas models appear to generate fewer false positives, whereas rural Texas models may be more sensitive to positive cases.

## Model-Level Interpretation

CatBoost, LightGBM, XGBoost, and Logistic Regression were the most consistently strong models across cohorts. They maintained relatively high accuracy, ROC AUC, and specificity while providing better F1 score and PR AUC than the transformer-based models.

HIR-M3 and SAINT occasionally achieved high recall, particularly in Texas. However, their lower precision, F1 score, PR AUC, and ROC AUC suggest that this high recall likely comes with more false-positive predictions. These models may be considered when minimizing false negatives is the primary objective, but they appear less balanced overall.

## Interpretation Considerations

The optimal model depends on the intended use case. If reducing false positives is most important, models with higher specificity and precision may be preferred. If missing a positive case is more costly, models with higher recall may be preferable.

CatBoost, LightGBM, Gradient Boosting, and XGBoost were the most consistently competitive models across both settings.

# Overall Task 1 Takeaway and Rationale for Task 2

## Overall Takeaway

Task 1 demonstrates that no single standalone model provides uniformly optimal performance across Nationwide versus Texas populations, Rural versus Urban settings, and condition-defined cohorts. Conventional models—particularly CatBoost, XGBoost, LightGBM, Gradient Boosting, and Logistic Regression—provided the strongest and most stable overall discrimination. However, their performance profiles varied by subgroup: Nationwide cohorts generally favored accuracy, ROC AUC, and specificity, whereas Texas and multimorbidity cohorts often favored recall, F1 score, and PR AUC.

The central finding is a persistent **recall-specificity trade-off**. Models that identify more positive cases often produce more false positives, while models with high specificity can miss positive cases. This trade-off becomes especially pronounced in rural, Texas, and multi-condition cohorts.

Therefore, Task 2 is needed to test whether combining a strong, balanced conventional model with HIR-M3 can improve sensitivity without unacceptably reducing specificity.

## Why Task 2 Is Needed

The Task 1 findings support moving beyond a “single best model” strategy for three reasons:

1. **Performance varies by cohort and subgroup.**  
   Nationwide models were generally more balanced and specific, whereas Texas models often had higher F1 score, PR AUC, precision, or recall. Rural and multimorbidity cohorts also showed different operating characteristics than Urban and single-condition cohorts.

2. **The preferred model depends on the clinical objective.**  
   If false positives are costly, a high-specificity model such as CatBoost, XGBoost, LightGBM, or Logistic Regression may be preferred. If missing a positive case is costly, a more sensitive approach is needed.

3. **HIR-M3 supplies complementary sensitivity.**  
   HIR-M3 was not the strongest standalone balanced classifier, but it repeatedly achieved very high recall in several cohorts. This suggests that HIR-M3 may identify positive cases missed by conventional models. Its weakness is low specificity and, in some subgroups, poor precision, F1 score, PR AUC, or ROC AUC.

Task 2 should therefore evaluate whether an ensemble can retain the discrimination and specificity of the conventional models while using HIR-M3 to recover missed positive cases.

## Why HIR-M3 Should Be Combined With Conventional Models

HIR-M3 should not be selected as the final standalone model based on Task 1. Its high recall often came with low specificity, low precision, and lower discrimination than CatBoost, XGBoost, or LightGBM. In some condition cohorts, HIR-M3 approached recall of 1.00 while specificity approached 0.00, indicating that it may classify nearly everyone as positive at the current threshold.

However, this same behavior makes HIR-M3 useful as an ensemble component.

HIR-M3 may provide value because:

- It is more sensitive to potential positive cases.
- It may capture nonlinear or interaction patterns that conventional models miss.
- It may improve recall in subgroups where conventional models favor specificity.
- It is particularly relevant when the clinical cost of a false negative is high.
- Its tendency toward false positives can potentially be moderated by combining it with a stronger conventional model.

The rationale is not that HIR-M3 is better alone; rather, it may provide **complementary predictions**. A conventional model can serve as the anchor for balanced discrimination and specificity, while HIR-M3 contributes additional sensitivity.

## Interpretation

The ensemble analysis supports different strategies by cohort:

- **Nationwide:** Favor conventional models. LightGBM alone is the leading option for ROC AUC and Brier score, CatBoost alone has the highest F1 score, and XGBoost + HIR-M3 at 90:10 is a secondary candidate because it achieved the highest PR AUC.

- **Generalizability:** A single universal HIR-M3 ensemble ratio is not supported. Nationwide results favor zero to minimal HIR-M3 weight, whereas Texas favors a HIR-M3-dominant XGBoost blend.

# Required Novel Metrics

## 1. False-Negative Burden Gap

For subgroup g:

FNR_g = FN_g / (TP_g + FN_g)

Calculate:
- Worst-group FNR: max(FNR_g)
- FNR gap: max(FNR_g) - min(FNR_g)
- FNR burden ratio: max(FNR_g) / min(FNR_g), only if denominator is nonzero and stable

Interpretation:
- Lower gap is better.
- A higher FNR identifies groups more likely to have true positive cases missed.
- Emphasize this metric because missed cases may represent clinically important harm.

## 2. False-Positive Burden Gap

For subgroup g:

FPR_g = FP_g / (FP_g + TN_g)

Calculate:
- Worst-group FPR: max(FPR_g)
- FPR gap: max(FPR_g) - min(FPR_g)
- FPR burden ratio: max(FPR_g) / min(FPR_g), only if denominator is nonzero and stable

Interpretation:
- Lower gap is better.
- A higher FPR identifies groups receiving disproportionate false-positive classifications.
- Report FPR and FNR gaps together to avoid shifting harm from missed cases to false alarms.

## 3. Worst-Group Calibration Error

For each subgroup, calculate:
- Brier score
- Expected Calibration Error (ECE), if predicted probabilities are available
- Calibration-in-the-large: mean(predicted probability) minus observed event rate
- Calibration slope, if feasible

Then calculate:
- Worst-group Brier score
- Worst-group ECE
- Maximum calibration gap between subgroups

Interpretation:
- Lower calibration error is better.
- ACT-parity thresholding changes the classification decision threshold but does not necessarily improve probability calibration.
- Flag models with good parity but poor subgroup calibration.

## 4. Net-Benefit Disparity

At each clinically meaningful threshold t, calculate subgroup net benefit:

NB_g(t) = TP_g / N_g - [FP_g / N_g] × [t / (1 - t)]

Calculate:
- Net benefit for each subgroup
- Worst-group net benefit
- Net-benefit gap: max(NB_g) - min(NB_g)
- Overall net benefit
- Comparison with “treat all” and “treat none” strategies, if applicable

Interpretation:
- Higher net benefit is better.
- Lower net-benefit gap indicates more equitable distribution of clinical benefit and harm.
- Include decision-curve plots by model and subgroup.

## 5. Fairness-Performance Stability Area

Evaluate each model across a clinically plausible threshold grid, for example 0.10 to 0.50 in increments of 0.01.

At each threshold, calculate:
- ACT-parity gap
- Recall
- Precision
- Specificity
- FNR gap
- FPR gap
- Net benefit

Define clinically acceptable criteria before analysis:
- ACT-parity gap <= [INSERT ACCEPTABLE GAP]
- Recall >= [INSERT MINIMUM RECALL]
- Precision >= [INSERT MINIMUM PRECISION]
- Optional: net benefit > 0

Calculate:

Fairness-Performance Stability Area =
Number of thresholds meeting all prespecified criteria / Total thresholds tested

Interpretation:
- Higher values indicate that a model is fair and clinically useful across a broad threshold range.
- Low values indicate that apparent parity is fragile and depends on a narrow threshold choice.

## 6. Transportability Gap

For each model and metric, compare Nationwide and Texas performance:

Transportability Gap = absolute value of(
    Metric_Nationwide - Metric_Texas

- Worst-group calibration error
- Net-benefit gap
- ACT-parity gap

Also report:
- Worst-cohort performance: minimum value across cohorts for metrics where higher is better
- Worst-cohort performance: maximum value across cohorts for error metrics where lower is better

Interpretation:
- Smaller transportability gaps indicate more stable performance across settings.

# Novel Evaluation Metrics

## Purpose

The following metrics are proposed as study-specific extensions to standard performance and fairness reporting. They are designed to quantify three questions not fully answered by ROC AUC, PR AUC, F1 score, Brier score, recall, specificity, or ACT-parity alone:

1. Does HIR-M3 add genuine value when combined with a conventional model?
2. Is fairness retained across a meaningful range of thresholds rather than at one selected threshold?
3. Does a model remain reliable when transported between Nationwide and Texas cohorts?

---

# 1. Incremental Ensemble Value

## Math

Incremental Ensemble Value (IEV) measures whether an ensemble outperforms **both** of its component models.

For a metric \(M\) where higher values are better, such as ROC AUC, PR AUC, F1 score, recall, specificity, or net benefit:

\[
IEV_M =
M_{\text{ensemble}} -
\max
\left(
M_{\text{base}},
M_{\text{HIR-M3}}
\right)
\]

For Brier score, where lower values indicate better calibration:

\[
IEV_{\text{Brier}} =
\min
\left(
\text{Brier}_{\text{base}},
\text{Brier}_{\text{HIR-M3}}
\right)
-
\text{Brier}_{\text{ensemble}}
\]

## Objective

The objective is to determine whether a Task 2 ensemble provides **complementary predictive value**, rather than simply matching or reproducing the performance of its better component.

## Use

Calculate IEV for the selected Task 2 ensembles:

- Nationwide: XGBoost + HIR-M3 at 90:10
- Texas: XGBoost + HIR-M3 at 30:70

Calculate IEV for:

- ROC AUC
- PR AUC
- F1 score
- Brier score
- Recall and specificity
- ACT-parity gap
- False-negative burden gap
- False-positive burden gap
- Worst-group calibration error
- Net benefit

Do not calculate IEV for degenerate ratios:

- A 100:0 ratio is identical to the base model.
- A 0:100 ratio is identical to HIR-M3.

## Interpretation

- \(IEV > 0\): the ensemble improves on both individual components.
- \(IEV = 0\): the ensemble offers no advantage beyond the stronger component.
- \(IEV < 0\): the ensemble is worse than at least one component.

A positive IEV in PR AUC or recall should still be interpreted alongside calibration, false-positive burden, and equity metrics. Ensemble learning is intended to exploit complementary model behavior, so IEV directly tests whether that goal was achieved. [web:269][web:272]

# 2. Fairness-Performance Stability Area

## Math

Fairness and clinical performance may be acceptable at one threshold but fail at nearby thresholds. The Fairness-Performance Stability Area (FPSA) measures the proportion of thresholds where a model meets all prespecified equity and clinical-performance requirements.

Let \(T\) be a clinically plausible threshold grid, for example:

\[
T = \{0.10, 0.11, 0.12, \ldots, 0.50\}
\]

At each threshold \(t\), define:

\[
I(t) =
\mathbb{1}
\left(
\text{ACT-Parity Gap}(t) \leq \delta
\right)
\cdot
\mathbb{1}
\left(
\text{Recall}(t) \geq r_{\min}
\right)
\cdot
\mathbb{1}
\left(
\text{Precision}(t) \geq p_{\min}
\right)
\]

Optionally, require positive clinical utility:

\[
I(t) =
I(t)
\cdot
\mathbb{1}
\left(
\text{Net Benefit}(t) > 0
\right)
\]

The FPSA is:

\[
FPSA =
\frac{1}{|T|}
\sum_{t \in T} I(t)
\]

Where:

- \(\delta\) is the maximum acceptable ACT-parity gap.
- \(r_{\min}\) is the minimum clinically acceptable recall.
- \(p_{\min}\) is the minimum clinically acceptable precision.
- \(\mathbb{1}(\cdot)\) equals 1 when the condition is satisfied and 0 otherwise.

## Objective

The objective is to determine whether a model has a **robust equitable operating range** rather than achieving fairness only at one narrow threshold.

## Use

For each Task 3 model:

1. Evaluate predictions across the prespecified threshold grid.
2. Calculate ACT-parity gap, recall, precision, specificity, FNR gap, FPR gap, and net benefit at every threshold.
3. Define the acceptable parity and clinical-performance criteria before viewing final model results.
4. Calculate FPSA separately for Nationwide and Texas.
5. Compare default-threshold and ACT-parity-adjusted versions of every model.

## Interpretation

- \(FPSA = 1\): the model meets all criteria at every tested threshold.
- \(FPSA = 0\): the model never satisfies all criteria at the same time.
- Higher values indicate that the model is both clinically useful and equitable across a wider operating range.
- Low FPSA indicates threshold fragility: the model appears fair only at a highly specific operating point.

This metric extends conventional parity reporting by incorporating threshold robustness, which is important because fairness measures and error trade-offs can change substantially as the classification threshold changes. Clinical AI fairness frameworks commonly assess subgroup differences in performance metrics, but FPSA additionally tests whether acceptable fairness is stable across decisions thresholds. [web:268][web:273]

# 3. Transportability Gap

## Math

The Transportability Gap quantifies how much a model’s performance, calibration, or fairness changes between the Nationwide and Texas cohorts.

For any metric \(M\):

\[
\text{Transportability Gap}_M =
\left|
M_{\text{Nationwide}} -
M_{\text{Texas}}
\right|
\]

For metrics where higher values are preferred, calculate the worst-cohort value:

\[
\text{Worst-Cohort Performance}_M =
\min
\left(
M_{\text{Nationwide}},
M_{\text{Texas}}
\right)
\]

For error metrics where lower values are preferred, calculate:

\[
\text{Worst-Cohort Error}_M =
\max
\left(
M_{\text{Nationwide}},
M_{\text{Texas}}
\right)
\]

## Objective

The objective is to evaluate **cross-cohort robustness**. A model should not be selected for broad use only because it performs well in one setting if its performance or fairness deteriorates in another setting.

## Use

Calculate a Transportability Gap for each final model and each relevant outcome:

- ROC AUC
- PR AUC
- Brier score
- Recall
- Specificity
- ACT-parity gap
- False-negative burden gap
- False-positive burden gap
- Worst-group calibration error
- Net-benefit gap
- Fairness-Performance Stability Area

Report both the gap and worst-cohort performance. For example, a model with a high average ROC AUC but poor ROC AUC in Texas should not be considered universally robust.

## Interpretation

- Smaller values indicate stronger cross-cohort stability.
- A gap of 0 indicates identical performance in Nationwide and Texas.
- Use this metric alongside worst-cohort performance: a small gap at uniformly poor performance is not desirable.
- Prefer models with a small transportability gap **and** strong performance in the weaker cohort.

Geographic or health-system differences can change data distributions, outcome prevalence, and model behavior, making robustness across deployment settings a central consideration for clinical AI. [web:268][web:272]
