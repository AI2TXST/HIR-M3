```python
df_TX = pd.read_csv('data/processed_final_mergedDF_TX.csv', low_memory=False)
df = pd.read_csv('data/processed_final_mergedDF.csv', low_memory=False)
```

```python
def basic_characteristics(df, name="df"):
    print(f"==== Basic characteristics: {name} ====\n")

    # Shape
    print("Shape (rows, columns):")
    print(df.shape)
    print("\n")

    # Number of unique beneficiaries
    print("Number of unique beneficiaries:")
    if "Beneficiary_ID" in df.columns:
        print(df["Beneficiary_ID"].nunique())
    else:
        print("Beneficiary_ID column not found.")
    print("\n")

    # Comorbidity prevalence (proportion of patients with each condition)
    comorb_cols = ["has_diabetes", "has_heart_failure", "has_hypertension"]
    existing_comorb = [c for c in comorb_cols if c in df.columns]
    if existing_comorb:
        print("Comorbidity prevalence (mean of indicator columns):")
        print(df[existing_comorb].mean())
    else:
        print("No comorbidity columns found among:", comorb_cols)
    print("\n")

    # Outcomes: readmission and death prevalence
    outcome_cols = ["ever_readmitted", "ever_deceased"]
    existing_outcomes = [c for c in outcome_cols if c in df.columns]
    if existing_outcomes:
        print("Outcome prevalence (proportion of patients):")
        print(df[existing_outcomes].mean())
    else:
        print("Outcome columns not found among:", outcome_cols)
    print("\n")

    print(f"==== End of basic characteristics: {name} ====\n")


# Use it for df_TX
basic_characteristics(df_TX, name="Texas cohort (df_TX)")
```

```python
basic_characteristics(df, name="All states cohort (df)")
```
